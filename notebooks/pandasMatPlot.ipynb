```python
import pandas as pd
import matplotlib.pyplot as plt

fileName = 'titanic_dataset.csv'
df = pd.read_csv(fileName)

```

```python
df2 = df.rename(columns={"Survived" : "survived"})

df2.head(5)

columnDataFrame = df.columns.to_list()


df['New Column'] = False
df.head(20)
```

```python
df = df.drop(columns=['New Column'])

df.head(1)
```

```python
plt.scatter(df.Age, df.Fare, color = "red", s=10)
plt.xlabel("Age")
plt.ylabel("Fare")
plt.title("Age vs Fare")
plt.ylim(0,0.5*df.Fare.max())
plt.grid()
plt.yticks([int(i) for i in range (0,int(0.5*df.Fare.max()),20)])
plt.savefig("Age_Fare.pdf")
```
