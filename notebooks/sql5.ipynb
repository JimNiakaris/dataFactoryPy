```python
import mysql.connector as mys
import pandas as pd

my_conn = mys.connect(
    host="localhost",
    user='root',
    password='123456',
    database='sakila'
)

query = """
select c.last_name, c.first_name, count(r.rental_id) as rentals
from customer c
left join rental r on r.customer_id = c.customer_id
group by c.customer_id, c.last_name, c.first_name
order by c.last_name, c.first_name
"""

my_cursor = my_conn.cursor()
my_cursor.execute(query)
rec = my_cursor.fetchall()


df = pd.DataFrame(rec, columns=["last_name", "first_name", "rentals"])

print(df)

alter = """
alter table language
modify last_update timestamp null
"""

insert = """
INSERT INTO language
(`language_id`,
`name`)
VALUES
(7,
"Portuguese"),
(8,
"latin"),
(9,
"Greek")
on duplicate key update name = values(name);
"""

my_cursor.execute(alter)

my_cursor.execute(insert)
my_conn.commit()

df = pd.read_sql("select * from language", my_conn)
print(df)


```
